```python
#print(len(organic_results))
def extract_paper_info(organic_results, df):
    papers = []
    for re in organic_results:
        #print(re)
        if "resources" in re.keys(): 
            #print(re['resources'])
            for resource in re['resources']:
                if 'file_format' in resource.keys() and resource['file_format'] == 'PDF':
                    dict = {}
                    #print(resource['link'])
                    if "title" in re.keys():
                        dict['title'] = re['title']
                    if "snippet" in re.keys():
                        dict['snippet'] = re['snippet']
                    dict['link'] = resource['link']
                    if df.empty or not df['title'].str.contains(re['title']).any():
                        papers.append(dict)
                    elif df['title'].str.contains(re['title']).any(): print("중복 ", re['title'])
                else : print(resource)

    if papers:
        new_df = pd.DataFrame(papers)
        df = pd.concat([df, new_df], ignore_index=True)
    return df
        
    
```

```python
def PDFtoTXT(url, output_filename):
    # 1. PDF 파일 다운로드: requests.get()를 사용해 PDF 파일을 가져옵니다.
    response = requests.get(url)
    
    # HTTP 응답 코드가 200이 아니면, 에러 메시지를 출력하고 함수 종료
    if response.status_code != 200:
        print(f"Error: HTTP 응답 코드 {response.status_code}  {output_filename}")
        return
    
    print("HTTP 응답 코드: 200 - PDF 파일 다운로드 성공")
    
    # 응답받은 바이너리 데이터를 BytesIO 객체에 저장하여 메모리 내 가상 파일로 생성합니다.
    pdf_file = BytesIO(response.content)
    
    # 2. PDF 파일 읽기: PyPDF2의 PdfReader를 사용하여 PDF 파일을 파싱합니다.
    try:
        # strict=False 옵션을 사용하면 비표준 PDF 형식에 대해서도 가능한 한 오류를 무시하며 읽어줍니다.
        reader = PdfReader(pdf_file, strict=False)
    except Exception as e:
        print("PDF 파일을 읽는 중 오류 발생:", e)
        return

    # 3. PDF의 모든 페이지에서 텍스트 추출: 각 페이지별로 텍스트를 추출하여 하나의 문자열에 누적합니다.
    full_text = ""
    for i, page in enumerate(reader.pages, start=1):
        text = page.extract_text()
        if text:
            # 페이지 구분을 위해 페이지 번호와 함께 텍스트를 추가합니다.
            full_text += f"==== Page {i} ====\n{text}\n\n"
        else:
            full_text += f"==== Page {i} ==== : 텍스트 추출 실패 또는 페이지가 비어 있음.\n\n"

    # 4. 추출된 텍스트를 txt 파일로 저장: open() 함수를 사용하여 파일을 쓰기 모드('w')로 엽니다.
    try:
        with open(output_filename, 'w', encoding='utf-8') as file:
            file.write(full_text)
        print(f"텍스트 추출 및 저장 완료: {output_filename}")
    except Exception as e:
        print("파일 저장 중 오류 발생:", e)
    
```

```python
import requests
from io import BytesIO
from pdfminer.high_level import extract_text
```

```python

def pdfminerExtract(url, output_filename):
    """
    주어진 URL에서 PDF 파일을 다운로드한 후,
    pdfminer.six를 사용하여 텍스트를 추출하고,
    추출된 텍스트를 output_filename으로 저장합니다.
    """
    # 1. PDF 파일 다운로드
    headers = {
    'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/115.0 Safari/537.36'
    }
    response = requests.get(url, headers=headers)
    if response.status_code != 200:
        print(f"Error: HTTP 응답 코드 {response.status_code}")
        return
    print("HTTP 응답 코드: 200 - PDF 파일 다운로드 성공")
    
    # 2. BytesIO 객체를 사용해 메모리 내에서 PDF 파일을 다룹니다.
    pdf_file = BytesIO(response.content)
    
    # 3. pdfminer.six의 extract_text 함수를 사용해 PDF 파일의 텍스트 추출
    try:
        text = extract_text(pdf_file)
    except Exception as e:
        print("텍스트 추출 중 오류 발생:", e)
        return
    
    # 4. 추출된 텍스트를 지정한 txt 파일로 저장
    try:
        with open(output_filename, 'w', encoding='utf-8') as f:
            f.write(text)
        print(f"텍스트 추출 및 저장 완료: {output_filename}")
    except Exception as e:
        print("파일 저장 중 오류 발생:", e)

# 함수 사용 예시
pdf_url = "https://necrocenenecrolandscaping.wordpress.com/wp-content/uploads/2018/01/braidotti_biopower-and-necropolitics.pdf"
output_file = "extracted_text_pdfminer.txt"
PDFtoTXT(pdf_url, output_file)

```

```python
import requests
from urllib.parse import urlparse, urlunparse

def bypass_403(url):
    # 기본 User-Agent 설정
    base_headers = {
        "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) "
                      "AppleWebKit/537.36 (KHTML, like Gecko) "
                      "Chrome/115.0 Safari/537.36"
    }
    
    results = []  # 각 시도 결과를 저장

    # 1. IP/Host 스푸핑 헤더 추가
    spoof_headers = base_headers.copy()
    spoof_headers.update({
        "X-Forwarded-For": "127.0.0.1",
        "X-Forwarded-Host": "localhost",
        "X-Real-IP": "127.0.0.1",
        "Referer": "http://localhost"
    })
    print(">> 1. 스푸핑 헤더 사용")
    try:
        response = requests.get(url, headers=spoof_headers)
        results.append(("스푸핑 헤더", response.status_code, response.text))
    except Exception as e:
        print("스푸핑 헤더 요청 중 오류:", e)
    
    # 2. Path 기반 우회: URL 경로에 특수 문자를 삽입하여 우회
    parsed = urlparse(url)
    if parsed.path:
        # 예를 들어, /secret → /;%09/secret 형태로 변경
        new_path = "/;%09" + parsed.path
        new_url = urlunparse((parsed.scheme, parsed.netloc, new_path, 
                              parsed.params, parsed.query, parsed.fragment))
        print(">> 2. Path 기반 우회 시도:", new_url)
        try:
            response = requests.get(new_url, headers=base_headers)
            results.append(("Path 기반 우회", response.status_code, response.text))
        except Exception as e:
            print("Path 기반 우회 요청 중 오류:", e)
    
    # 3. HTTP 메소드 변경: GET 대신 POST 방식으로 요청
    print(">> 3. POST 메소드 사용")
    try:
        response = requests.post(url, headers=base_headers)
        results.append(("POST 메소드", response.status_code, response.text))
    except Exception as e:
        print("POST 메소드 요청 중 오류:", e)
    
    # 4. URL 재작성 헤더 사용
    rewrite_headers = base_headers.copy()
    rewrite_headers.update({"X-Rewrite-URL": url})
    print(">> 4. X-Rewrite-URL 헤더 사용")
    try:
        response = requests.get(url, headers=rewrite_headers)
        results.append(("X-Rewrite-URL 헤더", response.status_code, response.text))
    except Exception as e:
        print("X-Rewrite-URL 헤더 요청 중 오류:", e)
    
    # 결과 출력: 각 방식별 상태 코드와 일부 응답 내용을 확인
    print("\n=== 우회 시도 결과 ===")
    for method, code, text in results:
        print(f"[{method}] 응답 코드: {code}")
        # 응답 내용의 앞 200자를 출력합니다.
        print(text[:200])
        print("-" * 50)
    
    return results
```
